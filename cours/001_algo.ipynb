## **I. Principes fondamentaux**

### ***Qu’est-ce que l’algorithmique ?***

L’**algorithmique** est la science qui étudie la résolution de problèmes en les décomposant en une suite d’actions élémentaires, appelées **instructions**. Un **algorithme** est une méthode systématique, une manière ordonnée de procéder, pour résoudre un problème, sans dépendre du langage de programmation utilisé.

**Définition simple :** un algorithme, c’est comme une recette de cuisine : si l’on suit les étapes dans le bon ordre, on obtient le résultat attendu.

### ***Historique de l’algorithmique***

**Origines anciennes :**

* **Euclide (vers 300 av. J.-C.)** : dans ses *Éléments*, il décrit une méthode pour calculer le PGCD (Plus Grand Commun Diviseur) de deux nombres, c’est l’un des plus anciens algorithmes encore en usage. ([Wikipédia](https://fr.wikipedia.org/wiki/Algorithme_d%27Euclide))
* **Al-Khwârizmî (9ᵉ siècle)** : Mathématicien perse, il a écrit des traités qui ont introduit et popularisé des méthodes de calcul avec les chiffres indo-arabes. Le mot “algorithme” dérive de son nom. ([Wikipédia](https://fr.wikipedia.org/wiki/Al-Khw%C3%A2rizm%C3%AE))

**Développement moderne :**

* Début du XXᵉ siècle, notamment **années 1930-1940** : Alan Turing et d’autres ont formalisé la notion de calcul, de machine de Turing, de ce qui est “calculable” ou non.
* Milieu XXᵉ siècle : apparition des ordinateurs, des langages de programmation, des premières implémentations concrètes d’algorithmes dans des machines.
* Aujourd’hui : l’algorithmique est partout — traitement de données, intelligence artificielle, cryptographie, optimisation, etc.

### ***L'algorithmique dans le développement***

**Dans le projet :**
- **Phase d'analyse** : Décomposer le problème en sous-problèmes
- **Phase de conception** : Créer la logique avant le code
- **Optimisation** : Améliorer les performances et l'efficacité
- **Maintenance** : Faciliter la compréhension et les modifications

**Dans l'équipe de développement :**
- **Communication** : Partager la logique indépendamment du langage
- **Documentation** : Expliquer la méthode de résolution
- **Collaboration** : Permettre à différents développeurs de comprendre l'approche
- **Validation** : Vérifier la logique avant l'implémentation

**Dans le programme :**
- **Structure** : Organiser le code de manière logique
- **Réutilisabilité** : Créer des solutions génériques
- **Performance** : Choisir les meilleures approches
- **Fiabilité** : Prévoir les cas d'erreur et les gérer

## **II. Syntaxe des éléments clés**

### ***Machine à état et universalité de Turing***

> 📌 **Culture générale** : cette section situe les fondements théoriques de l'informatique. Pas de panique s'il reste des zones d'ombre — retenez surtout l'idée qu'un ordinateur applique des règles simples, étape par étape.

**Concept de machine à état :**

Une **machine à état** est un modèle mathématique qui décrit le comportement d'un système. Elle possède :
- **États** : Les différentes situations possibles du système
- **Transitions** : Les règles qui déterminent le passage d'un état à un autre
- **Entrées** : Les données qui déclenchent les transitions
- **Sorties** : Les résultats produits par le système

**Exemple simple : Feu tricolore**
- **États** : Rouge, Vert, Orange  
- **Transitions** : Rouge → Vert → Orange → Rouge
- **Entrée** : Temps écoulé
- **Sortie** : Couleur affichée

**Universalité de Turing :**

**Alan Turing** a démontré qu'une machine simple avec des règles basiques peut résoudre n'importe quel problème calculable. Cette **machine de Turing** universelle comprend :

- **Ruban infini** : Mémoire pour stocker les données
- **Tête de lecture/écriture** : Lit et modifie les symboles
- **Unité de contrôle** : Applique les règles selon l'état actuel

**Implication pratique :** Tous nos ordinateurs modernes sont des "machines de Turing" - ils peuvent théoriquement résoudre tout problème calculable, mais avec des limitations pratiques (mémoire, temps).

### ***Les variables : définition, cycle de vie, types et enjeux***

**Définition d'une variable :**

Une **variable** est un espace de stockage dans la mémoire de l'ordinateur qui :
- **Contient une valeur** : Stocke des données (nombres, texte, etc.)
- **Possède un nom** : Identifiant pour accéder aux données
- **A un type** : Définit le genre de données stockées
- **Peut changer** : Sa valeur peut être modifiée pendant l'exécution

**Analogie :** Une variable est comme une boîte étiquetée dans laquelle on peut ranger et retrouver des objets.

**Cycle de vie d'une variable :**

1. **Déclaration** : On annonce l'existence de la variable
2. **Initialisation** : On lui donne une première valeur
3. **Utilisation** : On lit ou modifie sa valeur
4. **Destruction** : La mémoire est libérée (automatique ou manuelle)

**Les différents types de données :**

**Types simples (primitifs) :**
- **Entiers** : Nombres sans virgule (-3, 0, 42)
- **Réels** : Nombres à virgule (3.14, -2.5, 0.001)
- **Booléens** : Vrai ou Faux (True/False)
- **Caractères/Chaînes** : Lettres et texte ('a', "Bonjour")

**Types complexes (composés) :**
- **Tableaux/Listes** : Collections d'éléments
- **Structures/Objets** : Groupement de données liées
- **Dictionnaires** : Données avec clés et valeurs

**Enjeux des variables :**

**Enjeux techniques :**
- **Mémoire** : Chaque type occupe une place différente
- **Performance** : Certains types sont plus rapides à traiter
- **Sécurité** : Éviter les débordements et les erreurs de type

**Enjeux de développement :**
- **Lisibilité** : Noms explicites et types appropriés
- **Maintenance** : Code plus facile à comprendre et modifier  
- **Débogage** : Variables bien nommées facilitent la correction d'erreurs

### ***Structures de contrôle : les conditions***

**Qu'est-ce qu'une condition ?**

Une **condition** (ou structure conditionnelle) permet à un programme de prendre des décisions en fonction de certaines circonstances. C'est comme un carrefour où le programme choisit quelle route suivre.

**Principe général :**
```
SI (condition vraie) ALORS
    exécuter action A
SINON
    exécuter action B
FIN SI
```

**Types de conditions :**

**Condition simple (SI...ALORS) :**
- **Une seule possibilité** : Si la condition est vraie, on exécute l'action
- **Exemple** : Si il pleut, alors je prends un parapluie

**Condition alternative (SI...ALORS...SINON) :**
- **Deux possibilités** : Une action si vrai, une autre si faux
- **Exemple** : Si il pleut, alors je prends un parapluie, sinon je sors sans

**Conditions multiples (SI...SINON SI...SINON) :**
- **Plusieurs possibilités** : On teste plusieurs conditions successivement
- **Exemple** : 
  - Si il pleut → parapluie
  - Sinon si il neige → manteau chaud  
  - Sinon si il fait chaud → t-shirt
  - Sinon → veste légère

**Conditions imbriquées :**
- **Conditions dans des conditions** : Tests plus complexes
- **Exemple** : Si il fait beau ET si c'est le weekend, alors je vais au parc

**Éléments d'une condition :**
- **Expression à tester** : Ce qu'on veut vérifier (âge > 18)
- **Opérateurs de comparaison** : =, ≠, <, >, ≤, ≥
- **Opérateurs logiques** : ET, OU, NON
- **Résultat booléen** : La condition est soit vraie, soit fausse

### ***Structures de contrôle : les boucles***

**Qu'est-ce qu'une boucle ?**

Une **boucle** permet de répéter une série d'instructions plusieurs fois. C'est un mécanisme fondamental qui évite la répétition de code et permet de traiter des collections de données.

**Analogie :** Une boucle est comme une tâche répétitive - faire la vaisselle assiette par assiette jusqu'à ce qu'il n'y en ait plus.

**Types de boucles :**

**1. Boucle POUR (longueur connue) :**
- **Nombre de répétitions fixe** : On sait à l'avance combien de fois répéter
- **Structure** : POUR i DE 1 À 10 FAIRE ... FIN POUR
- **Exemple** : Afficher les nombres de 1 à 100

**2. Boucle TANT QUE (longueur inconnue) :**
- **Répétition conditionnelle** : On continue tant qu'une condition est vraie
- **Structure** : TANT QUE (condition) FAIRE ... FIN TANT QUE
- **Exemple** : Demander un mot de passe jusqu'à ce qu'il soit correct

**3. Boucle RÉPÉTER...JUSQU'À :**
- **Exécution garantie** : Au moins une exécution, puis test de sortie
- **Structure** : RÉPÉTER ... JUSQU'À (condition de sortie)
- **Exemple** : Lancer un dé jusqu'à obtenir un 6

**Éléments d'une boucle :**

**Variable de contrôle :**
- **Compteur** : Variable qui change à chaque itération
- **Indice** : Position actuelle dans la répétition

**Condition de sortie :**
- **Test d'arrêt** : Quand s'arrêter
- ⚠️ **Risque** : Boucle infinie si la condition n'est jamais fausse

**Corps de la boucle :**
- **Instructions répétées** : Ce qu'on fait à chaque tour
- **Modification de la variable** : Important pour éviter les boucles infinies

**Dangers des boucles :**
- **Boucle infinie** : Quand la condition de sortie n'est jamais atteinte
- **Performance** : Boucles trop longues ou mal optimisées
- **Erreurs de logique** : Mauvaise gestion des indices ou conditions

### ***Les mécanismes d'erreurs et les exceptions***

**Qu'est-ce qu'une erreur en algorithmique ?**

Une **erreur** (ou exception) est une situation anormale qui se produit pendant l'exécution d'un programme. C'est comme un obstacle imprévu qui empêche le programme de continuer normalement.

**Types d'erreurs :**

**Erreurs de syntaxe :**
- **Fautes de frappe** : Mots-clés mal écrits, parenthèses manquantes
- **Détection** : Découvertes avant l'exécution par l'analyseur
- **Exemple** : Oublier une parenthèse fermante

**Erreurs d'exécution (Runtime) :**
- **Pendant l'exécution** : Le programme se plante en cours de route
- **Causes courantes** : Division par zéro, fichier inexistant, mémoire insuffisante
- **Exemple** : Essayer d'ouvrir un fichier qui n'existe pas

**Erreurs de logique :**
- **Code incorrect** : Le programme fonctionne mais fait autre chose que prévu
- **Plus difficiles à détecter** : Pas de plantage, mais résultats faux
- **Exemple** : Calculer une moyenne en oubliant de diviser par le nombre d'éléments

**Gestion des exceptions :**

**Principe de base :**
```
ESSAYER
    instructions risquées
ATTRAPER (type d'erreur)
    traitement de l'erreur
FINALEMENT
    nettoyage (optionnel)
FIN ESSAYER
```

**Avantages de la gestion d'exceptions :**
- **Protection** : Le programme ne s'arrête pas brutalement
- **Récupération** : Possibilité de corriger ou contourner le problème
- **Information** : Messages d'erreur clairs pour l'utilisateur
- **Robustesse** : Programme plus stable et fiable

**Bonnes pratiques :**
- ⚠️ **Prévoir les cas d'erreur** : Anticiper ce qui peut mal se passer
- **Exceptions spécifiques** : Traiter différemment chaque type d'erreur
- **Messages clairs** : Expliquer ce qui s'est passé et comment corriger
- **Récupération élégante** : Permettre au programme de continuer si possible

### ***Les procédures (paramètres, valeurs de retour)***

**Qu'est-ce qu'une procédure ?**

Une **procédure** (aussi appelée fonction) est un bloc de code réutilisable qui accomplit une tâche spécifique. C'est comme une "mini-machine" qui prend des données en entrée et produit un résultat.

**Analogie :** Une procédure est comme une recette de cuisine : on lui donne des ingrédients (paramètres), elle suit les étapes, et produit un plat (résultat).

**Structure d'une procédure :**
```
PROCÉDURE nom_procédure(paramètre1, paramètre2, ...)
    DÉBUT
        instructions
        RETOURNER résultat (optionnel)
    FIN
```

**Les paramètres :**

**Paramètres d'entrée :**
- **Données fournies** : Informations nécessaires à la procédure
- **Exemple** : Pour calculer l'aire d'un rectangle → largeur et hauteur
- **Types** : Peuvent être de n'importe quel type (nombres, texte, listes...)

**Types de passage de paramètres :**
- **Par valeur** : Une copie de la donnée est transmise
- **Par référence** : L'adresse de la donnée est transmise
- **Impact** : Modification ou non de la donnée originale

**Valeurs de retour :**

**Fonction vs Procédure :**
- **Fonction** : Retourne une valeur (comme une calculatrice)
- **Procédure** : Effectue une action sans retourner de valeur (comme imprimer)

**Types de retour :**
- **Valeur simple** : Un nombre, un texte, un booléen
- **Valeur composite** : Une liste, un objet, plusieurs valeurs
- **Aucun retour** : La procédure fait juste une action

**Avantages des procédures :**
- **Réutilisabilité** : Écrire une fois, utiliser plusieurs fois
- **Lisibilité** : Code organisé et plus facile à comprendre
- **Maintenance** : Modifications centralisées
- **Tests** : Possibilité de tester chaque procédure séparément
- **Modularité** : Décomposition du problème en sous-problèmes

### ***Callbacks, pointeurs de fonctions, délégués***

> 📌 **Notions avancées** : ne vous inquiétez pas si tout n'est pas limpide à la première lecture — vous recroiserez ces concepts en pratique plus tard dans la formation.

**Concepts avancés des fonctions :**

Ces concepts permettent de traiter les fonctions comme des données, ouvrant la voie à des techniques de programmation très puissantes.

**Callbacks (Fonctions de rappel) :**

**Définition :**
- **Fonction passée en paramètre** : Une fonction A donne une fonction B à une fonction C
- **Exécution différée** : La fonction B sera appelée plus tard par C
- **Exemple concret** : "Appelle-moi quand tu as terminé"

**Utilisation pratique :**
- **Gestion d'événements** : Que faire quand l'utilisateur clique sur un bouton
- **Traitement asynchrone** : Actions à faire quand une tâche longue se termine
- **Personnalisation** : Permettre à l'utilisateur de définir un comportement spécifique

**Pointeurs de fonctions :**

**Concept :**
- **Adresse mémoire** : Un pointeur vers l'emplacement de la fonction en mémoire
- **Référence** : Permet d'appeler une fonction via son "adresse"
- **Flexibilité** : Choisir quelle fonction utiliser pendant l'exécution

**Exemple d'usage :**
```
fonction_calcul ← pointeur vers addition
SI (operation = "+") ALORS
    fonction_calcul ← pointeur vers addition
SINON SI (operation = "*") ALORS
    fonction_calcul ← pointeur vers multiplication
FIN SI
résultat ← fonction_calcul(a, b)
```

**Délégués :**

**Définition :**
- **Interface standardisée** : Manière élégante de gérer les pointeurs de fonctions
- **Sécurité** : Vérification des types de paramètres et de retour
- **Composition** : Possibilité de chaîner plusieurs fonctions

**Avantages de ces techniques :**
- **Polymorphisme** : Même interface, comportements différents
- **Extensibilité** : Facile d'ajouter de nouveaux comportements
- **Modularité** : Séparation claire des responsabilités
- **Flexibilité** : Comportement modifiable à l'exécution

**Applications courantes :**
- **Interfaces utilisateur** : Réaction aux clics, saisies clavier
- **Tri personnalisé** : Définir ses propres critères de tri
- **Programmation web** : Gérer les requêtes HTTP
- **Programmation événementielle** : Répondre aux changements d'état

## **III. Algorithmes courants et nécessaires**

### ***Exemples manuels : multiplication et Tours de Hanoï***

**Comprendre par l'exemple :**

Pour bien saisir l'algorithmique, commençons par des exemples que nous pouvons faire "à la main" avant de les programmer.

**Exemple 1 : Multiplication (méthode de l'école) :**

**Le problème :** Calculer 123 × 45

**L'algorithme manuel :**
```
    123
  ×  45
  -----
    615  (123 × 5)
   492   (123 × 4, décalé d'une position, soit 4920)
  -----
   5535
```

**Décomposition algorithmique :**
1. **Multiplier par chaque chiffre** : 5, puis 4
2. **Gérer les positions** : Décalage selon la position du chiffre  
3. **Additionner les résultats** : 615 + 4920 = 5535
4. **Répéter pour tous les chiffres**

**Exemple 2 : Les Tours de Hanoï :**

**Le problème :** Déplacer une pile de disques d'une tige à une autre.

**Règles :**
- **Jamais un gros disque sur un petit**
- **Un seul disque à la fois**
- **Objectif** : Reconstituer la pile sur la tige de destination

**L'algorithme récursif :**
1. **Déplacer n-1 disques** vers la tige intermédiaire
2. **Déplacer le plus gros disque** vers la destination
3. **Déplacer les n-1 disques** de l'intermédiaire vers la destination

**Pourquoi ces exemples ?**
- **Logique pure** : Pas de syntaxe de programmation
- **Visualisation** : On peut "voir" chaque étape
- **Récursivité naturelle** : Les Tours de Hanoï montrent l'auto-référence
- **Fondements** : Base pour comprendre les algorithmes plus complexes

### ***Solutions de représentations graphiques***

**L'importance de la visualisation :**

Les algorithmes complexes sont souvent difficiles à comprendre avec du texte seul. Les représentations graphiques nous aident à "voir" la logique et à mieux réfléchir.

**Types de représentations :**

**Organigrammes (Flowcharts) :**
- **Symboles standardisés** : 
  - Ovale/Ellipse = Début/Fin
  - Rectangle = Action/Traitement  
  - Losange = Décision/Condition
  - Cercle = Connecteur
- **Flèches** : Indiquent le sens d'exécution
- **Avantage** : Vision d'ensemble du processus

**Pseudo-code :**
- **Langage intermédiaire** : Entre le français et le code informatique
- **Mots-clés standardisés** : SI/ALORS, POUR, TANT QUE, etc.
- **Universel** : Compréhensible par tout programmeur
- **Exemple** :
  ```
  DÉBUT
      LIRE nombre
      SI nombre > 0 ALORS
          AFFICHER "Positif"
      SINON
          AFFICHER "Négatif ou nul"
      FIN SI
  FIN
  ```

**Arbres de décision :**
- **Structure arborescente** : Chaque nœud est une décision
- **Feuilles** : Résultats finaux
- **Branches** : Chemins possibles selon les conditions
- **Utilisation** : Systèmes experts, intelligence artificielle

**Schémas de données :**
- **Structures** : Tableaux, listes, arbres
- **Relations** : Comment les données sont liées
- **Évolution** : Comment les données changent pendant l'algorithme

**Avantages des représentations graphiques :**
- **Compréhension immédiate** : L'œil saisit plus vite qu'une lecture
- **Communication** : Partage d'idées avec l'équipe
- **Débogage** : Repérer les erreurs de logique
- **Documentation** : Expliquer le fonctionnement
- **Réflexion** : Aide à structurer la pensée algorithmique

### ***Séries de conditions et systèmes experts***

**Séries de conditions :**

**Qu'est-ce qu'une série de conditions ?**
- **Enchaînement de tests** : Plusieurs conditions liées logiquement
- **Objectif** : Prendre une décision complexe étape par étape
- **Évolution** : Chaque condition peut dépendre des précédentes

**Structure typique :**
```
SI (condition1) ALORS
    action1
SINON SI (condition2) ALORS
    action2  
SINON SI (condition3) ALORS
    action3
SINON
    action_par_défaut
FIN SI
```

**Exemple concret : Système de notes**
- Si note ≥ 16 → "Très bien"
- Sinon si note ≥ 14 → "Bien" 
- Sinon si note ≥ 12 → "Assez bien"
- Sinon si note ≥ 10 → "Passable"
- Sinon → "Insuffisant"

**Systèmes experts :**

**Définition :**
- **Intelligence artificielle** : Programme qui imite le raisonnement d'un expert humain
- **Base de connaissances** : Ensemble de règles et de faits
- **Moteur d'inférence** : Mécanisme qui applique les règles
- **Objectif** : Résoudre des problèmes dans un domaine spécialisé

**Structure d'un système expert :**
```
BASE DE FAITS :
- Le patient a de la fièvre
- Le patient tousse
- Le patient a mal à la gorge

RÈGLES :
- SI (fièvre ET toux ET mal de gorge) ALORS "suspicion grippe"
- SI (fièvre ET éruption cutanée) ALORS "suspicion varicelle"
- SI (toux ET essoufflement) ALORS "problème respiratoire"

CONCLUSION :
- Application des règles → Diagnostic probable
```

**Applications des systèmes experts :**
- **Médecine** : Aide au diagnostic
- **Droit** : Conseil juridique automatisé
- **Maintenance** : Diagnostic de pannes
- **Finance** : Évaluation de risques
- **Éducation** : Tuteurs intelligents

**Avantages :**
- **Précision** : Réflexion méthodique sans oubli
- **Rapidité** : Traitement instantané de nombreuses règles
- **Capitalisation** : Conservation de l'expertise humaine
- **Cohérence** : Mêmes décisions dans les mêmes situations

### ***Boucles de longueur prévisible - exemples multiples***

**Caractéristiques :**

**Définition :**
- **Nombre d'itérations connu** : On sait à l'avance combien de fois répéter
- **Variable de contrôle** : Généralement un compteur qui s'incrémente
- **Structure type** : POUR i DE début À fin FAIRE ... FIN POUR

**Exemples concrets :**

**Exemple 1 : Table de multiplication**
```
POUR i DE 1 À 10 FAIRE
    résultat ← 7 × i
    AFFICHER "7 × " + i + " = " + résultat
FIN POUR
```
- **Prévisible** : Exactement 10 répétitions
- **Compteur** : i varie de 1 à 10

**Exemple 2 : Calcul de moyenne sur un tableau**
```
somme ← 0
POUR i DE 0 À taille_tableau-1 FAIRE
    somme ← somme + tableau[i]
FIN POUR
moyenne ← somme / taille_tableau
```
- **Parcours complet** : On traite chaque élément
- **Longueur connue** : taille_tableau éléments

**Exemple 3 : Génération de motifs**
```
POUR ligne DE 1 À 5 FAIRE
    POUR étoile DE 1 À ligne FAIRE
        AFFICHER "*"
    FIN POUR
    AFFICHER nouvelle_ligne
FIN POUR
```
Résultat :
```
*
**
***
****
*****
```

**Exemple 4 : Calcul de factorielle**
```
factorielle ← 1
POUR i DE 1 À n FAIRE
    factorielle ← factorielle × i
FIN POUR
```
- **n répétitions** : Déterminé par la valeur d'entrée

**Avantages des boucles prévisibles :**
- **Estimation de performance** : Calcul exact du temps d'exécution
- **Gestion mémoire** : Allocation précise d'espace
- **Débogage facilité** : Pas de risque de boucle infinie
- **Parallélisation possible** : Traitement simultané sur plusieurs processeurs

**Applications courantes :**
- **Traitement de tableaux** : Parcours, calculs statistiques
- **Graphiques** : Génération de pixels, formes géométriques
- **Calculs mathématiques** : Séries, approximations numériques
- **Traitement de fichiers** : Quand la taille est connue

### ***Boucles de longueur imprévisible - exemples et dangers***

**Caractéristiques :**

**Définition :**
- **Nombre d'itérations inconnu** : Dépend de conditions qui évoluent
- **Condition de sortie** : TANT QUE (condition) ou RÉPÉTER...JUSQU'À
- ⚠️ **Risque majeur** : Boucle infinie si la condition ne change jamais

**Exemples concrets :**

**Exemple 1 : Recherche d'un élément**
```
trouvé ← FAUX
position ← 0
TANT QUE (NON trouvé ET position < taille_tableau) FAIRE
    SI tableau[position] = valeur_cherchée ALORS
        trouvé ← VRAI
    SINON
        position ← position + 1
    FIN SI
FIN TANT QUE
```
- **Imprévisible** : On peut trouver dès le 1er élément ou jamais

**Exemple 2 : Saisie utilisateur valide**
```
RÉPÉTER
    AFFICHER "Entrez un nombre entre 1 et 10 :"
    LIRE nombre
    SI (nombre < 1 OU nombre > 10) ALORS
        AFFICHER "Erreur ! Recommencez."
    FIN SI
JUSQU'À (nombre >= 1 ET nombre <= 10)
```
- **Dépend de l'utilisateur** : Nombre de tentatives variable

**Exemple 3 : Algorithme d'approximation**
```
approximation ← 1.0
erreur ← 1.0
TANT QUE (erreur > 0.001) FAIRE
    nouvelle_approximation ← (approximation + nombre/approximation) / 2
    erreur ← ABS(nouvelle_approximation - approximation)
    approximation ← nouvelle_approximation
FIN TANT QUE
```
- **Convergence** : S'arrête quand assez précis

**DANGERS et comment les éviter :**

**Boucle infinie :**
```
Mauvais exemple :
compteur ← 1
TANT QUE (compteur ≠ 10) FAIRE
    AFFICHER compteur
    compteur ← compteur + 2    ← Problème ! Saute 10
FIN TANT QUE
```

**Solutions de sécurité :**
- **Compteur limite** : Arrêter après N itérations max
- **Vérification des conditions** : S'assurer qu'elles peuvent devenir fausses
- **Variables de progression** : Contrôler que quelque chose change
- **Conditions de sortie multiples** : Plusieurs façons de sortir

**Exemple sécurisé :**
```
tentatives ← 0
trouvé ← FAUX
TANT QUE (NON trouvé ET tentatives < 1000) FAIRE
    ← traitement
    tentatives ← tentatives + 1
FIN TANT QUE

SI tentatives >= 1000 ALORS
    AFFICHER "Limite atteinte, arrêt forcé"
FIN SI
```

**Applications courantes :**
- **Recherches** : Dans des structures de données
- **Interfaces utilisateur** : Menus, saisies
- **Algorithmes itératifs** : Approximations, optimisations
- **Communications** : Attente de réponses réseau

### ***La récursivité - suppression de la récursivité***

**Qu'est-ce que la récursivité ?**

**Définition :**
- **Fonction qui s'appelle elle-même** : Auto-référence contrôlée
- **Décomposition** : Résoudre un problème en résolvant des versions plus petites
- **Condition d'arrêt** : Cas de base pour éviter l'infini
- **Élégance** : Solutions souvent plus courtes et naturelles

**Exemple classique : Factorielle**
```
FONCTION factorielle(n)
    SI n <= 1 ALORS                 ← Cas de base
        RETOURNER 1
    SINON
        RETOURNER n × factorielle(n-1)  ← Appel récursif
    FIN SI
FIN FONCTION
```

**Étapes d'exécution pour factorielle(4) :**
```
factorielle(4) = 4 × factorielle(3)
                = 4 × 3 × factorielle(2)
                = 4 × 3 × 2 × factorielle(1)
                = 4 × 3 × 2 × 1
                = 24
```

**Éléments essentiels de la récursivité :**
- **Cas de base** : Situation qui arrête la récursion
- **Appel récursif** : La fonction s'appelle avec des paramètres modifiés
- **Convergence** : Les paramètres doivent tendre vers le cas de base

**Suppression de la récursivité :**

**Pourquoi supprimer la récursivité ?**
- **Économie de mémoire** : Éviter l'empilement des appels
- **Performance** : Éliminer le coût des appels de fonction
- **Prévention de débordement** : Éviter "stack overflow"

**Techniques de transformation :**

**1. Utilisation d'une boucle :**
```
Récursif :
FONCTION factorielle_recursive(n)
    SI n <= 1 ALORS RETOURNER 1
    SINON RETOURNER n × factorielle_recursive(n-1)
FIN FONCTION

Itératif :
FONCTION factorielle_iterative(n)
    résultat ← 1
    POUR i DE 2 À n FAIRE
        résultat ← résultat × i
    FIN POUR
    RETOURNER résultat
FIN FONCTION
```

**2. Utilisation d'une pile (stack) :**
```
FONCTION factorielle_avec_pile(n)
    pile ← nouvelle pile vide
    résultat ← 1
    
    ← Empiler tous les nombres
    POUR i DE n À 2 FAIRE (en décroissant)
        EMPILER(pile, i)
    FIN POUR
    
    ← Dépiler et multiplier
    TANT QUE (pile non vide) FAIRE
        résultat ← résultat × DEPILER(pile)
    FIN TANT QUE
    
    RETOURNER résultat
FIN FONCTION
```

**Avantages et inconvénients :**

**Récursivité :**
- ✅ **Code élégant** et proche du problème mathématique
- ✅ **Facilité de compréhension** pour certains problèmes
- ❌ **Consommation mémoire** importante
- ❌ **Risque de débordement** pour les grandes valeurs

**Version itérative :**
- ✅ **Efficacité mémoire** et performance
- ✅ **Contrôle précis** de l'exécution
- ❌ **Code parfois plus complexe** à comprendre
- ❌ **Perte d'élégance** mathématique

### ***Algorithmes de tri***

**L'importance du tri :**

Le tri est l'une des opérations les plus fondamentales en informatique. Organiser des données facilite grandement leur recherche et leur traitement.

**Principe général :**
- **Entrée** : Liste d'éléments désorganisés
- **Sortie** : Même liste avec éléments dans un ordre défini
- **Critère** : Règle de comparaison (numérique, alphabétique, etc.)

**Algorithmes de tri principaux :**

**1. Tri par sélection :**
```
POUR i DE 0 À taille-2 FAIRE
    min_index ← i
    POUR j DE i+1 À taille-1 FAIRE
        SI tableau[j] < tableau[min_index] ALORS
            min_index ← j
        FIN SI
    FIN POUR
    ECHANGER tableau[i] et tableau[min_index]
FIN POUR
```
- **Principe** : Trouver le minimum et le placer au début
- **Complexité** : O(n²) - toujours, même dans le meilleur cas
- **Simple** à comprendre et implémenter

**2. Tri à bulles :**
```
POUR i DE 0 À taille-2 FAIRE
    échange_effectué ← FAUX
    POUR j DE 0 À taille-2-i FAIRE
        SI tableau[j] > tableau[j+1] ALORS
            ECHANGER tableau[j] et tableau[j+1]
            échange_effectué ← VRAI
        FIN SI
    FIN POUR
    SI NON échange_effectué ALORS
        SORTIR  ← Aucun échange : tableau déjà trié
    FIN SI
FIN POUR
```
- **Principe** : Les éléments "remontent" comme des bulles
- **Complexité** : O(n²) au pire, O(n) si déjà trié *(grâce au drapeau `échange_effectué` qui arrête la boucle)*
- **Lent** mais intuitif

**3. Tri par insertion :**
```
POUR i DE 1 À taille-1 FAIRE
    clé ← tableau[i]
    j ← i - 1
    TANT QUE (j >= 0 ET tableau[j] > clé) FAIRE
        tableau[j+1] ← tableau[j]
        j ← j - 1
    FIN TANT QUE
    tableau[j+1] ← clé
FIN POUR
```
- **Principe** : Comme trier des cartes à jouer
- **Complexité** : O(n²) au pire, O(n) si presque trié
- **Efficace** pour de petites listes

**4. Tri rapide (Quicksort) :**
```
FONCTION tri_rapide(tableau, début, fin)
    SI début < fin ALORS
        pivot ← partitionner(tableau, début, fin)
        tri_rapide(tableau, début, pivot-1)
        tri_rapide(tableau, pivot+1, fin)
    FIN SI
FIN FONCTION
```
- **Principe** : Diviser pour régner avec un pivot
- **Complexité** : O(n log n) en moyenne, O(n²) au pire
- **Très utilisé** en pratique

**5. Tri fusion (Mergesort) :**
```
FONCTION tri_fusion(tableau, début, fin)
    SI début < fin ALORS
        milieu ← (début + fin) / 2
        tri_fusion(tableau, début, milieu)
        tri_fusion(tableau, milieu+1, fin)
        fusionner(tableau, début, milieu, fin)
    FIN SI
FIN FONCTION
```
- **Principe** : Diviser puis fusionner
- **Complexité** : O(n log n) garanti
- **Stable** et prévisible

**Choix d'algorithme selon le contexte :**
- **Petites données** : Tri par insertion
- **Performance générale** : Tri rapide
- **Garantie de performance** : Tri fusion
- **Mémoire limitée** : Tri par sélection
- **Déjà presque trié** : Tri à bulles ou insertion

### ***Algorithmes de compression (avec et sans perte)***

**Principe de la compression :**

La compression consiste à réduire la taille des données tout en conservant leur utilité. C'est essentiel pour le stockage et la transmission d'informations.

**Pourquoi compresser ?**
- **Économie d'espace** : Moins de place sur disque dur
- **Transmission rapide** : Fichiers plus légers sur Internet  
- **Coûts réduits** : Moins de bande passante, moins de stockage
- **Performance** : Chargement plus rapide

**Types de compression :**

**1. Compression sans perte (Lossless) :**

**Principe :** On peut retrouver exactement les données originales.

**Algorithme de Huffman :**
```
1. COMPTER la fréquence de chaque caractère
2. CRÉER un arbre binaire basé sur les fréquences
3. ASSIGNER des codes courts aux caractères fréquents
4. REMPLACER chaque caractère par son code

Exemple : "AAABBC" 
- A (3 fois) → code : 0
- B (2 fois) → code : 10  
- C (1 fois) → code : 11
Résultat : "000101011" (9 bits au lieu de 48)
```

**Algorithme RLE (Run Length Encoding) :**
```
Principe : Remplacer les répétitions par "nombre + caractère"
Exemple : "AAAAABBBC" → "5A3B1C"
```

**Algorithme LZ77 :**
```
Principe : Référencer des séquences déjà vues
"ABABCABAB" → "AB(référence AB)C(référence ABAB)"
```

**2. Compression avec perte (Lossy) :**

**Principe :** On accepte de perdre des détails pour gagner plus d'espace.

**Compression d'images (JPEG) :**
- **Réduction des couleurs** : Moins de nuances
- **Échantillonnage** : Moins de pixels détaillés
- **Détails imperceptibles** : Suppression des détails que l'œil ne voit pas

**Compression audio (MP3) :**
- **Fréquences inaudibles** : Suppression des sons hors perception humaine
- **Masquage psychoacoustique** : Sons couverts par d'autres sons
- **Échantillonnage réduit** : Moins de précision temporelle

**Compression vidéo (H.264) :**
- **Images clés** : Images complètes de référence
- **Images différentielles** : Seulement les changements
- **Prédiction de mouvement** : Déduire les mouvements

**Comparaison des approches :**

**Sans perte :**
- ✅ **Qualité parfaite** préservée
- ✅ **Réversible** à 100%
- ❌ **Taux de compression** plus limité
- **Usage** : Textes, programmes, données critiques

**Avec perte :**
- ✅ **Très fort taux** de compression
- ✅ **Acceptable** pour multimédia
- ❌ **Dégradation** de la qualité
- ❌ **Irréversible**
- **Usage** : Photos, musique, vidéos

**Applications pratiques :**
- **ZIP, RAR** : Compression sans perte générale
- **JPEG, PNG** : Images
- **MP3, FLAC** : Audio
- **MP4, AVI** : Vidéo
- **GZIP** : Pages web

## **IV. Algorithmique et structures de données**

### ***Types et choix de structures***

**Qu'est-ce qu'une structure de données ?**

Une **structure de données** est une manière d'organiser et de stocker des informations dans un programme pour les utiliser efficacement. C'est comme choisir le bon type de rangement selon ce qu'on veut stocker.

**Structures de données fondamentales :**

**1. Tableau (Array) :**
- **Organisation** : Éléments consécutifs en mémoire
- **Accès** : Direct par index → tableau[3]
- **Vitesse** : Très rapide pour accéder à un élément
- **Taille** : Généralement fixe
- **Usage** : Données homogènes, accès fréquent par position

**2. Liste chaînée (Linked List) :**
- **Organisation** : Éléments reliés par des pointeurs
- **Accès** : Séquentiel, il faut parcourir
- **Ajout/Suppression** : Très efficace au début
- **Taille** : Dynamique, peut grandir/rétrécir
- **Usage** : Taille variable, insertions fréquentes

**3. Pile (Stack) :**
- **Principe** : LIFO (Last In, First Out) - comme une pile d'assiettes
- **Opérations** : PUSH (empiler), POP (dépiler)
- **Accès** : Seulement au sommet
- **Usage** : Gestion des appels de fonctions, annulation d'actions

**4. File (Queue) :**
- **Principe** : FIFO (First In, First Out) - comme une file d'attente
- **Opérations** : ENQUEUE (enfiler), DEQUEUE (défiler)
- **Accès** : Au début et à la fin
- **Usage** : Gestion des tâches, impression, requêtes web

**5. Arbre (Tree) :**
- **Organisation** : Structure hiérarchique avec racine et branches
- **Accès** : Recherche par parcours
- **Recherche** : Efficace si bien organisé
- **Usage** : Taxonomies, systèmes de fichiers, décisions

**6. Arbre binaire de recherche :**
- **Règle** : Gauche < Racine < Droite
- **Recherche** : O(log n) en moyenne
- **Insertion/Suppression** : Efficaces
- **Usage** : Bases de données, dictionnaires

**7. Graphe (Graph) :**
- **Organisation** : Nœuds reliés par des arêtes
- **Relations** : Connexions complexes entre éléments
- **Parcours** : Multiples chemins possibles
- **Usage** : Réseaux sociaux, GPS, Internet

**Critères de choix :**

**Performance :**
- **Vitesse d'accès** : Tableau > Liste chaînée
- **Vitesse d'insertion** : Liste chaînée > Tableau (au milieu)
- **Vitesse de recherche** : Arbre binaire > Liste

**Mémoire :**
- **Occupation** : Tableau compact, Liste avec pointeurs
- **Flexibilité** : Taille fixe vs dynamique

**Usage prévu :**
- **Accès fréquent par index** → Tableau
- **Insertions/suppressions fréquentes** → Liste ou Arbre
- **Traitement séquentiel** → Pile ou File
- **Recherches fréquentes** → Arbre binaire ou Table de hachage
- **Relations complexes** → Graphe

### ***Utilisation d'une structure depuis l'algorithme***

**Comment intégrer structures et algorithmes :**

**Principe fondamental :**
L'algorithme définit la **logique** (que faire), la structure de données définit l'**organisation** (comment stocker). Les deux doivent être cohérents pour une efficacité maximale.

**Exemples d'interaction :**

**1. Recherche dans un tableau :**
```
FONCTION recherche_lineaire(tableau, valeur)
    POUR i DE 0 À taille_tableau-1 FAIRE
        SI tableau[i] = valeur ALORS
            RETOURNER i
        FIN SI
    FIN POUR
    RETOURNER -1  ← Non trouvé
FIN FONCTION
```
- **Structure** : Tableau → accès direct par index
- **Algorithme** : Parcours linéaire
- **Complexité** : O(n)

**2. Recherche dans un arbre binaire :**
```
FONCTION recherche_arbre(nœud, valeur)
    SI nœud = NULL ALORS
        RETOURNER FAUX
    FIN SI
    SI nœud.valeur = valeur ALORS
        RETOURNER VRAI
    SINON SI valeur < nœud.valeur ALORS
        RETOURNER recherche_arbre(nœud.gauche, valeur)
    SINON
        RETOURNER recherche_arbre(nœud.droite, valeur)
    FIN SI
FIN FONCTION
```
- **Structure** : Arbre binaire ordonné
- **Algorithme** : Division récursive
- **Complexité** : O(log n) *(si l'arbre est équilibré ; O(n) au pire)*

**3. Gestion d'une pile pour calculatrice :**
```
FONCTION calculer_expression_postfixe(expression)
    pile ← nouvelle pile vide
    
    POUR CHAQUE token DANS expression FAIRE
        SI token est un nombre ALORS
            EMPILER(pile, token)
        SINON  ← C'est un opérateur
            b ← DEPILER(pile)
            a ← DEPILER(pile)
            résultat ← appliquer_opérateur(a, opérateur, b)
            EMPILER(pile, résultat)
        FIN SI
    FIN POUR
    
    RETOURNER DEPILER(pile)
FIN FONCTION
```
- **Structure** : Pile LIFO
- **Algorithme** : Traitement séquentiel avec mémorisation
- **Exemple** : "3 4 + 2 *" → ((3+4)*2) = 14

**4. Parcours de graphe (BFS) :**
```
FONCTION parcours_largeur(graphe, nœud_start)
    file ← nouvelle file vide
    visités ← ensemble vide
    
    ENFILER(file, nœud_start)
    MARQUER visités[nœud_start] ← VRAI
    
    TANT QUE file non vide FAIRE
        nœud_actuel ← DEFILER(file)
        TRAITER(nœud_actuel)
        
        POUR CHAQUE voisin DE nœud_actuel FAIRE
            SI voisin NON dans visités ALORS
                ENFILER(file, voisin)
                MARQUER visités[voisin] ← VRAI
            FIN SI
        FIN POUR
    FIN TANT QUE
FIN FONCTION
```
- **Structure** : Graphe + File FIFO
- **Algorithme** : Exploration niveau par niveau
- **Usage** : Plus court chemin, réseaux sociaux

**Bonnes pratiques d'association :**

**Cohérence structure-algorithme :**
- **Accès fréquent** : Utiliser des structures à accès direct
- **Modifications fréquentes** : Préférer les structures dynamiques
- **Recherches répétées** : Organiser pour optimiser la recherche

**Exemples de mauvais choix :**
- ❌ Recherche dans une liste non triée → Préférer un arbre ou tri préalable
- ❌ Insertions fréquentes dans un tableau → Préférer une liste chaînée
- ❌ Accès aléatoire dans une liste chaînée → Préférer un tableau

**Adaptation dynamique :**
```
SI (beaucoup de recherches ET peu de modifications) ALORS
    utiliser arbre_binaire_de_recherche
SINON SI (beaucoup de modifications ET peu de recherches) ALORS
    utiliser liste_chainée
SINON
    utiliser tableau_dynamique
FIN SI
```

### ***Pratiques recommandées***

**Pour les algorithmes :**

**Clarté et lisibilité :**
- **Noms explicites** : `calculer_moyenne()` plutôt que `calc()`
- **Commentaires pertinents** : Expliquer le "pourquoi", pas le "quoi"
- **Décomposition** : Diviser les problèmes complexes en sous-problèmes
- **Une responsabilité par fonction** : Chaque fonction fait une seule chose

**Robustesse :**
- **Vérification des entrées** : Valider les paramètres
- **Gestion des cas d'erreur** : Prévoir ce qui peut mal se passer
- **Conditions limites** : Tester les valeurs extrêmes
- **Conditions de sortie** : Éviter les boucles infinies

**Performance :**
- **Choisir le bon algorithme** : Adapter à la taille des données
- **Économie de mémoire** : Réutiliser les variables si possible
- **Éviter les calculs répétés** : Mémoriser les résultats intermédiaires
- **Mesurer, ne pas deviner** : Profiler avant d'optimiser

**Pour les structures de données :**

**Choix adapté :**
- **Analyser l'usage** : Quelles opérations sont les plus fréquentes ?
- **Estimer la taille** : Combien d'éléments maximum ?
- **Prévoir l'évolution** : Les données vont-elles croître ?
- **Équilibrer** : Temps vs espace mémoire

**Encapsulation :**
- **Interface claire** : Masquer la complexité interne
- **Protection des données** : Empêcher les accès directs dangereux
- **Opérations cohérentes** : Maintenir l'intégrité des données
- **Testabilité** : Faciliter les tests unitaires

**Exemples de bonnes pratiques :**

**Code bien structuré :**
```
FONCTION tri_fusion(tableau)
    SI taille(tableau) <= 1 ALORS
        RETOURNER tableau  ← Cas de base
    FIN SI
    
    milieu ← taille(tableau) / 2
    gauche ← tri_fusion(tableau[0...milieu-1])
    droite ← tri_fusion(tableau[milieu...fin])
    
    RETOURNER fusionner(gauche, droite)
FIN FONCTION

FONCTION fusionner(gauche, droite)
    ← Fusion des deux parties triées
    ← Une seule responsabilité !
FIN FONCTION
```

**Gestion d'erreurs :**
```
FONCTION diviser(a, b)
    SI b = 0 ALORS
        LEVER EXCEPTION "Division par zéro impossible"
    FIN SI
    RETOURNER a / b
FIN FONCTION
```

**Interface propre :**
```
STRUCTURE Pile
    PRIVÉ éléments : tableau
    PRIVÉ sommet : entier
    
    PUBLIQUE empiler(valeur)
    PUBLIQUE dépiler() : valeur
    PUBLIQUE est_vide() : booléen
    PUBLIQUE taille() : entier
FIN STRUCTURE
```

**Réutilisabilité :**
- **Fonctions génériques** : Paramètres pour adapter le comportement
- **Modules/Bibliothèques** : Regrouper les fonctions liées
- **Documentation** : Expliquer l'usage et les paramètres
- **Tests** : Assurer la fiabilité

**Maintenance :**
- **Code autodocumenté** : Le code explique sa fonction
- **Versions** : Historique des modifications
- **Métriques** : Suivre la complexité et performance
- **Revues de code** : Validation par les pairs

### ***Introduction à la Théorie des Graphes***

**Qu'est-ce qu'un graphe ?**

Un **graphe** est une structure mathématique composée de **nœuds** (ou sommets) reliés par des **arêtes** (ou liens). C'est un modèle puissant pour représenter des relations entre éléments.

**Analogies concrètes :**
- **Plan de ville** : Intersections (nœuds) reliées par des rues (arêtes)
- **Réseau social** : Personnes (nœuds) reliées par des amitiés (arêtes)  
- **Internet** : Ordinateurs (nœuds) reliés par des connexions (arêtes)
- **Organigramme** : Postes (nœuds) reliés par des relations hiérarchiques

**Éléments d'un graphe :**

**Nœuds (Sommets) :**
- **Identité** : Chaque nœud a un identifiant unique
- **Données** : Peut contenir des informations (nom, coordonnées, etc.)
- **Degré** : Nombre d'arêtes connectées à ce nœud

**Arêtes (Liens) :**
- **Connexion** : Relie deux nœuds
- **Poids** : Peut avoir une valeur (distance, coût, intensité)
- **Direction** : Peut être orientée ou non

**Types de graphes :**

**Graphe non orienté :**
- **Relations symétriques** : Si A est relié à B, alors B est relié à A
- **Exemple** : Amitiés, routes bidirectionnelles
- **Représentation** : A——B (ligne simple)

**Graphe orienté (Digraphe) :**
- **Relations asymétriques** : A peut pointer vers B sans que B pointe vers A
- **Exemple** : Suivre sur Twitter, pages web avec liens
- **Représentation** : A→B (flèche)

**Graphe pondéré :**
- **Arêtes avec valeurs** : Chaque connexion a un poids
- **Exemple** : Distances entre villes, coût des trajets
- **Représentation** : A——(15km)——B

**Graphe simple vs multigraphe :**
- **Simple** : Au maximum une arête entre deux nœuds
- **Multigraphe** : Plusieurs arêtes possibles entre mêmes nœuds

**Propriétés importantes :**

**Connexité :**
- **Graphe connexe** : On peut aller de n'importe quel nœud vers n'importe quel autre
- **Composantes connexes** : Groupes de nœuds reliés entre eux

**Cycles :**
- **Cycle** : Chemin qui revient au point de départ
- **Arbre** : Graphe connexe sans cycles
- **Forêt** : Collection d'arbres

**Applications concrètes :**
- **GPS** : Plus court chemin entre deux points
- **Internet** : Routage des données
- **LinkedIn** : Recommandations de contacts
- **E-commerce** : Recommandations de produits
- **Réseaux électriques** : Distribution optimale
- **Bioinformatique** : Relations entre gènes
- **Gestion de projet** : Dépendances entre tâches

**Algorithmes de parcours de graphes :**

Pour exploiter un graphe, il faut d'abord savoir le **parcourir**, c'est-à-dire visiter ses nœuds de manière méthodique.

**1. Parcours en largeur (BFS - Breadth-First Search) :**
- **Principe** : Explorer les nœuds niveau par niveau, en s'appuyant sur une **file** (FIFO)
- **Usage** : Plus court chemin en nombre d'arêtes, amis d'amis dans un réseau social
- 📌 Le pseudo-code complet (`parcours_largeur`) a été présenté dans la section *Utilisation d'une structure depuis l'algorithme* (exemple 4)

**2. Parcours en profondeur (DFS - Depth-First Search) :**
```
FONCTION parcours_profondeur(graphe, nœud)
    MARQUER nœud comme visité
    TRAITER(nœud)

    POUR CHAQUE voisin DE nœud FAIRE
        SI voisin non visité ALORS
            parcours_profondeur(graphe, voisin)
        FIN SI
    FIN POUR
FIN FONCTION
```
- **Principe** : Suivre un chemin jusqu'au bout avant de revenir en arrière (récursion ou **pile**)
- **Usage** : Sortir d'un labyrinthe, détection de cycles, exploration exhaustive

**3. Algorithme de Dijkstra (plus court chemin pondéré) :**
```
FONCTION dijkstra(graphe, source)
    POUR CHAQUE nœud n DU graphe FAIRE
        distances[n] ← INFINI
    FIN POUR
    distances[source] ← 0
    non_visités ← tous les nœuds du graphe

    TANT QUE non_visités non vide FAIRE
        u ← nœud de non_visités avec la plus petite distance
        RETIRER u de non_visités

        POUR CHAQUE voisin v DE u FAIRE
            alt ← distances[u] + poids(u, v)
            SI alt < distances[v] ALORS
                distances[v] ← alt
            FIN SI
        FIN POUR
    FIN TANT QUE

    RETOURNER distances
FIN FONCTION
```


**Caractéristiques :**
- **Plus court chemin depuis un nœud source**
- **Gère les poids positifs**
- **Structure** : File à priorité
- **Complexité** : O((V + E) log V)

**Autres algorithmes importants :**

**4. Détection de cycles :**
- **DFS modifié** : Repérer les arêtes vers des nœuds en cours de visite
- **Usage** : Vérifier les dépendances circulaires

**5. Tri topologique :**
- **Ordonner les tâches** : Respecter les dépendances
- **Exemple** : Ordre de compilation de modules
- **Algorithme** : DFS avec pile de résultats

**6. Composantes fortement connexes :**
- **Groupes de nœuds** : Tous atteignables mutuellement
- **Usage** : Analyse de réseaux sociaux, web

**Applications pratiques :**

**Navigation GPS :**
- **Graphe** : Intersections et routes
- **Algorithme** : Dijkstra ou A*
- **Objectif** : Plus court chemin

**Réseaux sociaux :**

### ***La notion de complexité algorithmique (notation grand O)***

**Pourquoi mesurer la complexité ?**

La **complexité algorithmique** permet de prédire comment un algorithme va se comporter quand la taille des données augmente. C'est essentiel pour choisir le bon algorithme selon le contexte.

**Analogie :** C'est comme comparer des moyens de transport :
- **À pied** : Lent mais pas de préparation
- **Voiture** : Plus rapide mais il faut du carburant  
- **Avion** : Très rapide sur longue distance mais préparation longue

**Types de complexité :**

**1. Complexité temporelle :**
- **Temps d'exécution** : Combien de temps prend l'algorithme
- **En fonction de n** : Taille des données d'entrée
- **Objectif** : Prédire les performances

**2. Complexité spatiale :**
- **Mémoire utilisée** : Combien d'espace occupe l'algorithme
- **Croissance** : Comment l'usage mémoire évolue

**Notation Big O (Grand O) :**

**Principe :** On s'intéresse au **comportement asymptotique** - comment l'algorithme se comporte quand n devient très grand.

**Classes de complexité courantes :**

**O(1) - Complexité constante :**
- **Temps fixe** : Même durée peu importe la taille
- **Exemple** : Accéder à un élément d'un tableau par son index
```
élément ← tableau[5]  ← Toujours la même vitesse
```

**O(log n) - Complexité logarithmique :**
- **Croissance lente** : Doubler n ajoute juste une étape
- **Exemple** : Recherche binaire, arbres équilibrés
- **Intuition** : On élimine la moitié à chaque étape

**O(n) - Complexité linéaire :**
- **Proportionnel** : Doubler n double le temps
- **Exemple** : Parcourir un tableau, recherche linéaire
```
POUR chaque élément DANS tableau FAIRE  ← n opérations
    traiter élément
FIN POUR
```

**O(n log n) - Complexité quasi-linéaire :**
- **Efficace** : Meilleurs tris comparatifs
- **Exemple** : Tri fusion, tri rapide (cas moyen)
- **Intuition** : Diviser pour régner

**O(n²) - Complexité quadratique :**
- **Croissance rapide** : Doubler n quadruple le temps
- **Exemple** : Tri à bulles, algorithmes naïfs
```
POUR i DE 0 À n-1 FAIRE
    POUR j DE 0 À n-1 FAIRE  ← n × n = n² opérations
        comparer tableau[i] et tableau[j]
    FIN POUR
FIN POUR
```

**O(2ⁿ) - Complexité exponentielle :**
- **Explosion** : Très vite impraticable
- **Exemple** : Certains problèmes de force brute
- ⚠️ **Danger** : Même avec n=50, c'est ingérable

**Comparaison visuelle pour n=1000 :**
- O(1) : **1 opération**
- O(log n) : **10 opérations**
- O(n) : **1 000 opérations**
- O(n log n) : **10 000 opérations**
- O(n²) : **1 000 000 opérations**
- O(2ⁿ) : **Plus d'atomes dans l'univers !**

**Comment analyser la complexité :**

**1. Identifier les boucles :**
```
POUR i DE 0 À n FAIRE        ← O(n)
    POUR j DE 0 À n FAIRE    ← O(n)
        opération simple     ← O(1)
    FIN POUR
FIN POUR
Total : O(n) × O(n) × O(1) = O(n²)
```

**2. Prendre le pire cas :**
- **Scénario pessimiste** : Quand l'algorithme est le plus lent
- **Garantie** : Performance minimale assurée

**3. Ignorer les constantes :**
- O(3n + 5) = O(n)
- O(n²/2 + n) = O(n²)
- **Raison** : Pour de grandes valeurs, seul le terme dominant compte

**Applications pratiques :**
- **Recherche Google** : Optimisations pour des milliards de pages
- **Trading haute fréquence** : Microsecondes critiques
- **Jeux vidéo** : 60 FPS obligent à optimiser
- **Apps mobiles** : Batterie et processeur limités

### ***Table de hachage***

**Principe de la table de hachage :**

Une **table de hachage** (hash table) est une structure de données qui permet d'associer des **clés** à des **valeurs** avec un accès très rapide. C'est comme un annuaire téléphonique ultra-efficace.

**Analogie :** Imaginez une bibliothèque où chaque livre aurait un code unique qui indique directement sur quelle étagère et à quelle position le trouver, sans avoir à chercher.

**Composants principaux :**

**1. Fonction de hachage :**
- **Transformation** : Convertit une clé en index numérique
- **Objectif** : Répartir les clés uniformément
- **Exemple simple** :
```
hash("Alice") = somme_ascii("Alice") % taille_table
hash("Alice") = (65+108+105+99+101) % 100 = 78
```

**2. Tableau (buckets) :**
- **Stockage** : Chaque case contient une paire clé-valeur
- **Accès direct** : tableau[hash(clé)]
- **Taille fixe** : Déterminée à la création

**Fonctionnement de base :**

**Insertion :**
```
FONCTION insérer(clé, valeur)
    index ← fonction_hash(clé)
    table[index] ← (clé, valeur)
FIN FONCTION
```

**Recherche :**
```
FONCTION rechercher(clé)
    index ← fonction_hash(clé)
    RETOURNER table[index].valeur
FIN FONCTION
```

**Le problème des collisions :**

**Qu'est-ce qu'une collision ?**
- **Même index** : Deux clés différentes produisent le même hash
- **Exemple** : hash("Alice") = hash("Bob") = 78
- ⚠️ **Inévitable** : Statistiquement, cela arrive toujours

**Solutions aux collisions :**

**1. Chaînage (Separate Chaining) :**
```
Chaque case contient une liste de paires clé-valeur

table[78] → [("Alice", "123-456"), ("Bob", "789-012")]

FONCTION rechercher(clé)
    index ← fonction_hash(clé)
    POUR chaque (k,v) DANS table[index] FAIRE
        SI k = clé ALORS RETOURNER v
    FIN POUR
    RETOURNER NULL
FIN FONCTION
```
- **Avantage** : Simple à implémenter
- **Inconvénient** : Mémoire supplémentaire pour les listes

**2. Adressage ouvert (Open Addressing) :**
```
En cas de collision, chercher la prochaine case libre

FONCTION insérer(clé, valeur)
    index ← fonction_hash(clé)
    TANT QUE table[index] occupée FAIRE
        index ← (index + 1) % taille_table  ← Sondage linéaire
    FIN TANT QUE
    table[index] ← (clé, valeur)
FIN FONCTION
```
- **Avantage** : Pas de mémoire supplémentaire
- **Inconvénient** : Plus complexe, effet de clustering

**Qualités d'une bonne fonction de hachage :**

**Uniformité :**
- **Répartition égale** : Les clés sont distribuées uniformément
- **Éviter les clusters** : Pas de zones surchargées

**Déterminisme :**
- **Même résultat** : La même clé donne toujours le même hash
- **Reproductible** : Important pour la cohérence

**Vitesse :**
- **Calcul rapide** : O(1) pour calculer le hash
- **Pas trop complexe** : Équilibrer qualité et rapidité

**Exemples de fonctions de hachage :**

**Pour les chaînes de caractères :**
```
FONCTION hash_string(chaine)
    hash ← 0
    POUR chaque caractère c DANS chaine FAIRE
        hash ← hash * 31 + code_ascii(c)
    FIN POUR
    RETOURNER hash % taille_table
FIN FONCTION
```

**Performance :**

**Cas idéal :**
- **Recherche** : O(1)
- **Insertion** : O(1)
- **Suppression** : O(1)

**Cas dégradé (nombreuses collisions) :**
- **Toutes opérations** : O(n) - devient comme une liste

**Applications concrètes :**
- **Dictionnaires** : Mots et définitions
- **Bases de données** : Index pour accès rapide
- **Cache web** : Stockage temporaire de pages
- **Jeux vidéo** : Gestion rapide des objets par ID
- **Compilateurs** : Table des symboles
- **Sécurité** : Vérification de mots de passe
