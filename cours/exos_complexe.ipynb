1. Analyse de texte

Ecrire une fonction qui prend un texte en entrée et retourne :

    * le nombre total de mots
    * les 5 mots les plus fréquents
    * les mots uniques (n'apparaissent qu'une fois)

```python
from collections import Counter
import re

def analyser_texte(p_texte : str) -> dict :
    '''
    compter le nombre total de mot, les mots les plus fréquents et les mots uniques 
    '''
    mots = re.findall(r"\b[\wÀ-ÿ]+\b", p_texte.lower())
    compteur = Counter(mots)

    quantite = len(mots)
    plus_frequents = compteur.most_common(5)
    mots_uniques = []
    for mot, frequence in compteur.items():
        if frequence == 1:
            mots_uniques.append(mot)

    return {
        "total_mots" : quantite,
        "5_mots_les_plus_frequents" : plus_frequents,
        "mots_uniques": mots_uniques
    }


with open("exos.md", "r", encoding="utf-8") as f:
    contenu = f.read()
    # print(contenu)

resultat = analyser_texte(contenu)
print(resultat)
```

2. Gestionnaire de stock

```python
stock_produits = {
    "chaise" : 25,
    "velo" : 19,
    "canapé" : 2,
}

def ajouter_produit(produit: list):
    '''
    Ajoute un produit au stock existant
    '''
    if produit[0] in stock_produits.keys(): # on verifie si l'item à ajouter n'est pas déjà présent
        stock_produits[produit[0]] += 1
    else:
        stock_produits[produit[0]] = produit[1]

def retirer_produit(produit: str):
    '''
    Retire un produit des stocks'''
    if produit in stock_produits.keys():
        if stock_produits[produit] > 1: # Si le stock est positif, on décrémente de 1
            stock_produits[produit] -= 1
        else: # sinon on retire directement le produit.
            del stock_produits[produit]
            print(f"le produit {produit} a été retiré")
    else: # Si le produit n'est déjà plus dans les stocks
        print("ce produit ne figure pas dans les stocks")

def afficher_stock():
    for cle, valeur in stock_produits.items():
        print(f"{valeur} {cle}(s) en stock")
```

```python
prod1 = ("table",12)
prod2 = ("velo",)
ajouter_produit(prod1)
print(stock_produits)
ajouter_produit(prod2)
print(stock_produits)
```
