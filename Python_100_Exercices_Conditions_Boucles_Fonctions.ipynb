### 1. Conditions simples 

1. Écrire un programme qui vérifie si un nombre donné est supérieur à 10

```python
nombre = int(input("Entrez un nombre : "))
if nombre > 10:
    print(f"{nombre} est supérieur à 10.")
elif nombre < 10:
    print(f"{nombre} est inférieur à 10.")
else:
    print(f"{nombre} est égal à 10.")
```

7. Vérifier si un nombre est divisible par 3 ou par 5

```python
nombre = int(input("Entrez un nombre : "))
if nombre % 3 == 0:
    print(f"{nombre} est divisible par 3.")
elif nombre % 5 == 0:
    print(f"{nombre} est divisible par 5.")
elif nombre % 3 == 0 and nombre % 5 == 0:
    print(f"{nombre} est divisible par 3 et par 5.")
elif nombre % 5 == 0:
    print(f"{nombre} n'est divisible ni par 3, ni par 5.")
```

9. Créer un programme qui vérifie si une lettre est une voyelle ou une consonne.

19. Créer un programme qui vérifie si une chaîne de caractères est un palindrome.

20. Écrire un programme qui vérifie si un nombre est à la fois positif, pair et supérieur à 100.

```python
nombre = 101

positif = nombre > 0
pair = nombre % 2 == 0
superieur = nombre > 100

if positif and pair and superieur:
    print(f"{nombre} est à la fois positif, pair et supérieur à 100")
else:
    if positif:
        print(f"{nombre} est positif")
    if pair:
        print(f"{nombre} est pair")
    if superieur:
        print(f"{nombre} est superieur à 100")
```

24. Demander à l'utilisateur un nombre de 1 à 10 et afficher la table de multiplication. 

```python
nombre = -1
saisi_ok = nombre > 0 and nombre < 10
print(saisi_ok)
while not saisi_ok:
    nombre = int(input("Entrez un nombre entre 1 et 10"))
    saisi_ok = nombre > 0 and nombre < 10
    
print(nombre)
for i in range(1,11):
    print(f"{i} * {nombre} = {i * nombre}")
```

```python
nombre = -1
saisi_ok = False
print(saisi_ok)
while not saisi_ok:
    nombre = int(input("Entrez un nombre entre 1 et 10"))
    if nombre > 0 and nombre < 10:
        saisie_ok = True

print(nombre)
for i in range(1, 11):
    print(f"{i} * {nombre} = {i * nombre}")
```
